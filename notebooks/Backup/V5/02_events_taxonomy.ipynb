## What this notebook does

Adds 8 event indicator columns to the reconstructed dataset.
An **event** is a binary or categorical condition at the time of each received packet.  
Example: *"Was CO₂ above 700 ppm when this packet arrived?"*

| Event | What it measures |
|-------|-------------------|
| E1 | CO₂ tier |
| E2 | CO₂ rising (first-order delta) |
| E3 | Office hours |
| E4 | PM2.5 tier |
| E5 | Temperature tier |
| E6 | Humidity tier |
| E7 | Spreading factor |
| E8 | ESP tier |
| Burst loss | Descriptive classification of loss-episode length, not a modelled event |

Notebook 03 correlates these events with `PDR_link` (outage intervals excluded).

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams.update(
    {
        'font.family': 'serif',
        'font.size': 12,
        'axes.titlesize': 13,
        'axes.labelsize': 13,
        'xtick.labelsize': 12,
        'ytick.labelsize': 12,
        'legend.fontsize': 11,
        'figure.dpi': 150,
        'savefig.dpi': 300,
        'savefig.bbox': 'tight',
        'axes.spines.top': False,
        'axes.spines.right': False,
        'axes.grid': True,
        'grid.alpha': 0.3,
        'grid.linestyle': '--',
    }
)
C = {
    'blue': '#0077BB',
    'orange': '#EE7733',
    'green': '#009988',
    'red': '#CC3311',
    'purple': '#AA3377',
    'navy': '#004488',
    'grey': '#BBBBBB',
    'brown': '#882255',
    'teal': '#44AA99',
    'yellow': '#DDCC77',
}
DEVICE_COLORS = [C['blue'], C['orange'], C['green'], C['red'], C['purple'], C['navy']]
TOA_MS = {7: 71.9, 8: 133.6, 9: 246.8, 10: 452.6, 11: 987.1, 12: 1810.4}
FIG_DIR = Path('../figures/2')
FIG_DIR.mkdir(exist_ok=True)


def save_fig(fig, name):
    # fig.savefig(FIG_DIR/f'{name}.pdf')
    fig.savefig(FIG_DIR / f'{name}.png')
    print(f'  Saved: figures/{name}')
    plt.show()


print('Figures settings ready.')


def donut_chart(ax, values, colors, labels, small_threshold=3.0, radius=1.0):
    """
    Donut chart with automatic label placement, no title (captions carry
    that in the thesis). Wedges at or above small_threshold percent get an
    inline label; thinner wedges get an external label on a leader line,
    staggered vertically so adjacent thin wedges don't overlap.
    """
    wedges, _ = ax.pie(
        values,
        colors=colors,
        startangle=90,
        radius=radius,
        wedgeprops=dict(width=0.42, edgecolor='white', linewidth=1.5),
    )

    small_above = 0
    small_below = 0
    for wedge, val, lab in zip(wedges, values, labels):
        ang = np.radians((wedge.theta2 + wedge.theta1) / 2)
        x, y = np.cos(ang), np.sin(ang)

        if val >= small_threshold:
            ax.text(
                x * 0.78 * radius, y * 0.78 * radius, f'{val:.1f}%',
                ha='center', va='center', fontsize=12, fontweight='bold', color='white',
            )
        else:
            side = 1 if x >= 0 else -1
            if y >= 0:
                y_stag = 1.08 + 0.17 * small_above
                small_above += 1
            else:
                y_stag = -(1.08 + 0.17 * small_below)
                small_below += 1
            ax.annotate(
                f'{lab}\n{val:.2f}%',
                xy=(x * 0.82 * radius, y * 0.82 * radius),
                xytext=(1.45 * radius * side, y_stag * radius),
                ha='left' if side > 0 else 'right', va='center',
                fontsize=10, fontweight='bold',
                arrowprops=dict(arrowstyle='-', color='#666666', lw=1),
            )
    return wedges
```

## 1 · CO₂ Tier  *(§6.1)*

CO₂ rises linearly with the number of people present — human metabolism produces ~200 ml CO₂/min at rest.  
Outdoor ambient is ~420 ppm. Weekday mean in this building is ~576 ppm vs ~457 ppm on weekends.

| Event | Definition |
|-------|-----------|
| E1 | CO₂ tier: background (<500) / moderate (500–700) / high (>700 ppm) |

```python
# E1: CO2 tier — 500 ppm = outdoor ambient ceiling, 700 ppm = moderate occupancy onset
df['e1_co2_tier'] = pd.cut(
    df['co2'], bins=[0, 500, 700, 10_000], labels=['background', 'moderate', 'high']
)

print('CO2 tier distribution (E1):')
print(
    df['e1_co2_tier']
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
    .astype(str)
    .add('%')
    .to_string()
)
print(
    f'\nCO2 mean overall         : {df["co2"].mean():.0f} ppm  |  std: {df["co2"].std():.0f} ppm'
)
```

```python
# Fig 02-A — CO₂ Tier Distribution (E1)
tiers = ['background', 'moderate', 'high']
xlabs = ['Background', 'Moderate', 'High']
pcts = df['e1_co2_tier'].value_counts(normalize=True).reindex(tiers).fillna(0) * 100
colors = [C['green'], C['orange'], C['red']]

fig, ax = plt.subplots(figsize=(6, 6))
donut_chart(ax, pcts.values, colors, xlabs)
ax.legend([plt.Rectangle((0, 0), 1, 1, color=c) for c in colors], xlabs,
          loc='center', fontsize=11, frameon=False)

fig.tight_layout()
save_fig(fig, '02_a1_co2_tier_distribution')
```

## 2 · CO₂ Rising  *(§6.2)*

| Event | Definition |
|-------|-----------|
| E2 | CO₂ rising: delta > 15 ppm per 60s |

```python
# E2: CO2 rising — delta > 15 ppm/60s
# 15 ppm chosen to exceed Sensirion SCD41 sensor noise (~±10 ppm)
df['e2_co2_rising'] = (df.groupby('device_id')['co2'].diff() > 15).astype(int)

print(
    f'CO2 rising intervals (E2): {df["e2_co2_rising"].sum():,}  ({df["e2_co2_rising"].mean() * 100:.1f}% of all intervals)'
)
```

## 3 · Office Hours  *(§6.3)*

Converted to `Europe/Berlin` local time — handles both DST transitions  
(October 2024 and March 2025) automatically.

| Event | Definition |
|-------|-----------|
| E3 | Office hours — weekday AND 08:00–17:59 |

```python
# local time — required for correct hour/weekday extraction across DST boundaries
local = df['time'].dt.tz_convert('Europe/Berlin')
df['hour'] = local.dt.hour
df['dayofweek'] = local.dt.dayofweek  # 0=Monday, 6=Sunday

# E3: office hours — weekday AND 08:00–17:59
df['is_weekday'] = (df['dayofweek'] < 5).astype(int)

df['e3_office_hours'] = ((df['is_weekday'] == 1) & df['hour'].between(8, 17)).astype(int)

print(f'Office hours (E3): {df["e3_office_hours"].mean() * 100:.1f}%')
```

```python
# Fig 02-B — Temporal Event Distribution (E3)
oh = df['e3_office_hours'].mean() * 100
vals = [oh, 100 - oh]
labs = ['Office Hours', 'Off-Peak']
colors = [C['red'], C['blue']]

fig, ax = plt.subplots(figsize=(5.5, 5.5))
donut_chart(ax, vals, colors, labs)
ax.legend([plt.Rectangle((0, 0), 1, 1, color=c) for c in colors], labs,
          loc='center', fontsize=11, frameon=False)

fig.tight_layout()
save_fig(fig, '02_b_temporal_distributions')
```

## 7 · Spreading Factor Event  *(§6.7)*

SF directly controls Time-on-Air (ToA) — the duration a packet occupies the channel.  
Under LoRaWAN's ALOHA protocol, longer ToA = wider collision window = higher loss probability.

ADR was disabled. SF rotated in a fixed cycle: SF12→SF11→SF10→SF9→SF8→SF7.  
SF11/SF12 blocks are truncated to 4 packets by duty cycle enforcement (§5.2.1).

| SF | ToA (ms) | Collision window vs SF7 |
|----|---------|------------------------|
| 7 | 71.9 | baseline |
| 8 | 133.6 | 1.9× |
| 9 | 246.8 | 3.4× |
| 10 | 452.6 | 6.3× |
| 11 | 987.1 | 13.7× |
| 12 | 1810.4 | 25.2× |

| Event | Definition |
|-------|-----------|
| E7   | SF categorical: 7 / 8 / 9 / 10 / 11 / 12 |

```python
# E7: individual SF value
TOA_MS = {7: 71.9, 8: 133.6, 9: 246.8, 10: 452.6, 11: 987.1, 12: 1810.4}
df['e7_sf'] = df['SF'].astype(int)

print('SF Distribution (E7):')
for sf in [7, 8, 9, 10, 11, 12]:
    n = (df['e7_sf'] == sf).sum()
    print(f'  SF{sf} (ToA={TOA_MS[sf]:7.1f}ms): {n:,} ({n / len(df) * 100:.1f}%)')
```

```python
# Fig 02-E — SF Distribution (E7)
sfs = [7, 8, 9, 10, 11, 12]
sf_labs = [f'SF{sf}' for sf in sfs]
sf_pcts = [float(df['e7_sf'].eq(sf).mean() * 100) for sf in sfs]
sf_colors = [C['green'], C['blue'], C['navy'], C['orange'], C['red'], C['brown']]

fig, ax = plt.subplots(figsize=(7.5, 7.5))
donut_chart(ax, sf_pcts, sf_colors, sf_labs)
ax.legend([plt.Rectangle((0, 0), 1, 1, color=c) for c in sf_colors], sf_labs,
          loc='center', fontsize=10, frameon=False, ncol=2)

fig.tight_layout()
save_fig(fig, '02_e_sf_distribution')
```

## 8 · ESP Event  *(§6.8)*

ESP (effective signal power) is only available for received packets — never for lost ones.  
E8 uses the ESP of the packet received immediately before the loss interval  
as a proxy for channel state at the time of loss.

| Event | Definition |
|-------|-----------|
| E8   | ESP tier: poor / moderate / strong (threshold source under review — see notebook 03) |

```python
# E8: ESP tier — collapsed to 3 analytical tiers from ThingPark 6-band classification
# poor    = critical + very_weak  (below -120 dBm)
# moderate = weak + fair          (-120 to -100 dBm)
# strong  = good + excellent      (above -100 dBm)
df['e8_esp_tier'] = pd.cut(
    df['esp'], bins=[-300, -120, -100, 0], labels=['poor', 'moderate', 'strong']
)

print("ESP tier distribution (E8) — 3 analytical tiers:")
print(
    df['e8_esp_tier']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
    .reindex(['poor', 'moderate', 'strong'])
    .to_string()
)
print(f"\nRaw counts:")
print(df['e8_esp_tier'].value_counts().reindex(['poor', 'moderate', 'strong']).to_string())
```

```python
# Fig 02-F — ESP Event Distribution (E8) — 3 analytical tiers
et = ['poor', 'moderate', 'strong']
et_labs = ['Poor', 'Moderate', 'Strong']
et_pcts = [float((df['e8_esp_tier'].astype(str) == t).mean() * 100) for t in et]
et_colors = [C['red'], C['orange'], C['green']]

fig, ax = plt.subplots(figsize=(6.5, 6.5))
donut_chart(ax, et_pcts, et_colors, et_labs)
ax.legend([plt.Rectangle((0, 0), 1, 1, color=c) for c in et_colors], et_labs,
          loc='center', fontsize=11, frameon=False)

fig.tight_layout()
save_fig(fig, '02_f_signal_quality_distribution')
```
